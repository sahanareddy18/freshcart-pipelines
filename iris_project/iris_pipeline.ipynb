# Mage Iris Pipeline

The project in `metadata.yaml` wires a loader, two tested transformers, and two downstream exporters. This notebook runs the same blocks locally so the model outputs can be inspected without starting Mage. Use Mage CLI/UI for the assignment's orchestration screenshots and runtime-variable evidence.

```python
import sys
import importlib
from pathlib import Path

project_root = Path.cwd().parent if Path.cwd().name == "iris_project" else Path.cwd()
sys.path.insert(0, str(project_root))

from iris_project.data_loaders.load_iris import load_data
from iris_project.transformers.validate_iris import transform as validate_iris
from iris_project.transformers.prepare_data import transform as prepare_data
from iris_project.transformers.train_model import train_model, test_model_accuracy
from iris_project.data_exporters.evaluate_model import evaluate_model, test_evaluation_output

save_results_module = importlib.import_module("iris_project.data_exporters.save_results")
save_results = importlib.reload(save_results_module).save_results

iris = validate_iris(load_data())
prepared = prepare_data(iris)
training = train_model(prepared, max_depth=3)
test_model_accuracy(training)
evaluation = evaluate_model(training)
test_evaluation_output(evaluation)
saved_path = save_results(training, output_path="artifacts/iris_predictions.csv")
print(f"Accuracy: {training['accuracy']:.4f}")
print("Confusion matrix:")
print(evaluation["confusion_matrix"])
print("Metrics:")
print(evaluation["metrics"].to_string(index=False))
print(f"Predictions saved to {saved_path}")
```
